```python
# Uplift Causal Model - Final Version
# Causal Uplift Modeling pour Kolecto – Démonstration

import numpy as np
import pandas as pd
import pickle
import os
from doubleml import DoubleMLData, DoubleMLPLR
from lightgbm import LGBMRegressor
import matplotlib.pyplot as plt

# Ensure output directories exist
os.makedirs('../results/figures/uplift', exist_ok=True)
os.makedirs('../results/metrics/uplift', exist_ok=True)
```

```python
# 1. Charger les données
with open('../data/processed/churn_data.pkl', 'rb') as f:
    data = pickle.load(f)

X_train = data['X_train_scaled']
y_train = data['y_train']
X_test = data['X_test_scaled']
```

```python
# 2. IMPORTANT : En vraie production, on aurait une variable "treatment" (appel CX fait ou non)
# Ici, on simule pour démonstration (20% des utilisateurs ont reçu un appel)
np.random.seed(42)
treatment = np.random.binomial(1, 0.2, size=len(y_train))
```

```python
# 3. DoubleML – Estimation de l’effet moyen (ATE)
dml_data = DoubleMLData.from_arrays(X_train, y_train, treatment)
learner = LGBMRegressor(n_estimators=200, max_depth=6, learning_rate=0.05, random_state=42)

dml_plr = DoubleMLPLR(dml_data, ml_l=learner, ml_m=learner, n_folds=5)
print("Fitting DoubleML pour ATE...")
dml_plr.fit()

print("\n=== EFFET MOYEN D'UNE INTERVENTION CX (ATE) ===")
print(f"Gain moyen estimé : +{dml_plr.coef[0]:.3f} points de conversion")
try:
    confint = dml_plr.confint()
    if hasattr(confint, 'iloc'):
        ci_low = confint.iloc[0, 0]
        ci_high = confint.iloc[0, 1]
    else:
        ci_low = confint[0][0]
        ci_high = confint[0][1]
    print(f"IC 95% : [{ci_low:.3f}, {ci_high:.3f}]")
except:
    print("Could not calculate CI")
```

```python
# 4. T-Learner – Uplift individuel (CATE)
mask_treat = (treatment == 1)
m0 = LGBMRegressor(n_estimators=200, max_depth=6, learning_rate=0.05, random_state=42)
m1 = LGBMRegressor(n_estimators=200, max_depth=6, learning_rate=0.05, random_state=42)

m0.fit(X_train[~mask_treat], y_train[~mask_treat])
m1.fit(X_train[mask_treat], y_train[mask_treat])

uplift_pred = m1.predict(X_test) - m0.predict(X_test)
```

```python
# 5. Base probability (ton LightGBM existant)
# Corrected path to the model
model_path = '../results/models/lightgbm/lightgbm.pkl'
if os.path.exists(model_path):
    with open(model_path, 'rb') as f:
        lgb_model = pickle.load(f)
    # Check if model has predict_proba
    if hasattr(lgb_model, 'predict_proba'):
        base_prob = lgb_model.predict_proba(X_test)[:, 1]
    else:
        # Fallback if it's a regressor or pipeline
        base_prob = lgb_model.predict(X_test)
else:
    print(f"Warning: Model not found at {model_path}. Training a temporary classifier.")
    from lightgbm import LGBMClassifier
    lgb_model = LGBMClassifier(n_estimators=200, max_depth=6, learning_rate=0.05, random_state=42)
    lgb_model.fit(X_train, y_train)
    base_prob = lgb_model.predict_proba(X_test)[:, 1]
```

```python
# 6. Classification CX
df_uplift = pd.DataFrame({
    'base_prob': base_prob,
    'uplift': uplift_pred
})

df_uplift['category'] = pd.cut(df_uplift['uplift'], 
                               bins=[-np.inf, 0.0, 0.1, 0.3, np.inf],
                               labels=['Lost Cause', 'Do-not-disturb', 'Sleeping Dog', 'Persuadable'])

print("\n=== RÉPARTITION DES UTILISATEURS ===")
print(df_uplift['category'].value_counts(normalize=True).round(3))

print("\n=== RECOMMANDATIONS CX ===")
print("• Appeler uniquement les 'Persuadable' (uplift > 0.3)")
print("• Ne pas déranger les 'Do-not-disturb'")
print("→ Gain estimé : +10-20 % d’efficacité des interventions CX")
```

```python
# 7. Visualisation
plt.figure(figsize=(10, 6))
# Handle potential NaN in plotting
valid_idx = ~df_uplift['uplift'].isna()
scatter = plt.scatter(df_uplift.loc[valid_idx, 'base_prob'], 
                      df_uplift.loc[valid_idx, 'uplift'], 
                      c=df_uplift.loc[valid_idx, 'uplift'], 
                      cmap='RdYlGn', alpha=0.7)
plt.colorbar(scatter, label='Uplift estimé')
plt.axhline(0.1, color='orange', linestyle='--')
plt.axhline(0.3, color='green', linestyle='--')
plt.xlabel('Probabilité de conversion (modèle LightGBM)')
plt.ylabel('Uplift estimé (gain avec appel CX)')
plt.title('Identification des utilisateurs "Persuadables"')
plt.savefig('../results/figures/uplift/uplift_scatter.png')
print("Plot saved to ../results/figures/uplift/uplift_scatter.png")
plt.show()

# Save metrics for consistency
import json
metrics = {
            "model": "Causal Uplift",
        "ate": float(dml_plr.coef[0]),
    "segmentation": df_uplift['category'].value_counts(normalize=True).to_dict(),
    "description": "Final Causal Uplift Model Analysis"
}
try:
    confint = dml_plr.confint()
    if hasattr(confint, 'iloc'):
         metrics["ate_conf_interval"] = [float(confint.iloc[0, 0]), float(confint.iloc[0, 1])]
    else:
         metrics["ate_conf_interval"] = [float(confint[0][0]), float(confint[0][1])]
except:
    pass

with open('../results/metrics/uplift/metrics.json', 'w') as f:
    json.dump(metrics, f, indent=4)
```
